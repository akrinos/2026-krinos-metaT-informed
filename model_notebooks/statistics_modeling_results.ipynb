```r
for_prob_test = infill_dates %>%
    dplyr::mutate(DateTimeForm=as.Date(paste0(format(as.Date(DateTime,format="%Y-%m-%d"), 
                                                     "%Y-%m"), "-01")))%>%
    dplyr::group_by(taxon,DateTimeForm) %>% 
    dplyr::summarize(abund=mean(abund,na.rm=TRUE))

all_stats = data.frame()
for (curr_group in unique(for_prob_test$taxon)) {
    frame_for_test = for_prob_test %>% dplyr::filter(taxon==curr_group) %>%
        dplyr::arrange(DateTimeForm)
    stats_test = mk.test(frame_for_test$abund)
    curr_fr = data.frame(statistic = stats_test$statistic,
               organism = curr_group,
               p_value = stats_test$p.value,
               total_abundance = sum(frame_for_test$abund),
               avg_abundance = mean(frame_for_test$abund))
    all_stats = all_stats %>% dplyr::bind_rows(curr_fr)
    print(curr_group)
    print(mk.test(frame_for_test$abund))
               
}
```
